# 객체지향프로그래밍(실) 과제 - Map 활용 및 Encapsulation

- 이름: 김도연

```java
import java.util.HashMap;
import java.util.Map;

public class ModifiableLibrary {
    private String name;
    private Map<String, String> books; // <도서 제목, 저자>

    // 생성자
    public ModifiableLibrary(String name) {
        this.name = name;
        this.books = new HashMap<>();
    }

    // Getter & Setter (Encapsulation)
    public String getName() {
        return name;
    }

    public void setName(String name) {
        this.name = name;
    }

    public Map<String, String> getBooks() {
        return new HashMap<>(books);
    }

    // 도서 추가
    public void addBook(String title, String author) {
        books.put(title, author);
    }

    // 도서 삭제
    public boolean removeBook(String title) {
        if (books.containsKey(title)) {
            books.remove(title);
            return true;
        }
        return false;
    }

    // 도서 검색
    public String findAuthor(String title) {
        return books.getOrDefault(title, "도서를 찾을 수 없습니다.");
    }

    // toString 오버라이딩
    @Override
    public String toString() {
        return "ModifiableLibrary{" +
                "name='" + name + '\'' +
                ", totalBooks=" + books.size() +
                ", books=" + books +
                '}';
    }
}
```

```java
// ModifiableLibrary 동작 테스트
ModifiableLibrary library = new ModifiableLibrary("한남대학교 중앙도서관");

// 도서 추가
library.addBook("자바의 정석", "남궁성");
library.addBook("클린 코드", "로버트 C. 마틴");
library.addBook("이펙티브 자바", "조슈아 블로크");

// 상태 출력 (Override)
System.out.println(library);

// 검색 테스트
System.out.println("조회 결과 (클린 코드): " + library.findAuthor("클린 코드"));

// 삭제 테스트
library.removeBook("자바의 정석");
System.out.println("삭제 후 상태: " + library);
```
