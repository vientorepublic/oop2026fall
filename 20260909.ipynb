```java
import java.util.Arrays;

record StudentScores(String name, int[] scores) {
    public void updateFirstScore(int newScore) {
        if (this.scores != null && this.scores.length > 0) {
            this.scores[0] = newScore;
        }
    }
}

int[] scores = {80, 90, 95};
StudentScores student = new StudentScores("홍길동", scores);
System.out.println("변경 전 점수: " + Arrays.toString(student.scores()));

student.updateFirstScore(100);
System.out.println("변경 후 점수: " + Arrays.toString(student.scores()));
```
