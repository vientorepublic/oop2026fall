```java
import java.util.Map;

Map<String, List<String>> routeMap = Map.ofEntries(
    Map.entry("서울", List.of("원주", "천안")),
    Map.entry("원주", List.of("평창", "안동")),
    Map.entry("평창", List.of("강릉")),
    Map.entry("강릉", List.of()),
    Map.entry("안동", List.of()),
    Map.entry("천안", List.of("대전", "공주")),
    Map.entry("대전", List.of("대구", "익산")),
    Map.entry("공주", List.of("익산")),
    Map.entry("대구", List.of("부산")),
    Map.entry("부산", List.of()),
    Map.entry("익산", List.of("광주", "전주")),
    Map.entry("광주", List.of("목포")),
    Map.entry("목포", List.of()),
    Map.entry("전주", List.of("여수")),
    Map.entry("여수", List.of())
);
```

```java
boolean hasPath(String orig, String dest) {
    // 출발지와 도착지가 같으면 경로 존재하면 true
    if (orig.equals(dest)) {
        return true;
    }
    
    List<String> neighbors = routeMap.get(orig);
    if (neighbors == null || neighbors.isEmpty()) {
        return false;
    }
    
    // 인접한 지점 중 하나라도 dest까지 노선이 연결된 경우가 있다면 true
    for (String next : neighbors) {
        if (hasPath(next, dest)) {
            return true;
        }
    }
    
    return false;
}
```

```java
System.out.println(hasPath("서울", "목포")); // true (서울 -> 천안 -> 공주 -> 익산 -> 광주 -> 목포)
System.out.println(hasPath("부산", "서울")); // false (역방향 하행선 불가)
System.out.println(hasPath("원주", "여수")); // false (연결되지 않은 다른 노선 가지)
```
