```java
// 1. 제너릭 및 인터페이스 구현 (주제 2, 3)

import java.util.ArrayList;
import java.util.List;

// 1. 인터페이스 정의 (결제 수단의 공통 기능)
interface PaymentMethod {
    void processPayment(double amount);
    String getPaymentType();
}

// 2. 추상 클래스 정의 (상품의 공통 속성 및 메서드)
abstract class Product {
    private String name;
    private double price;

    public Product(String name, double price) {
        this.name = name;
        this.price = price;
    }

    public String getName() {
        return name;
    }

    public double getPrice() {
        return price;
    }

    // 하위 클래스에서 각 상품 타입에 맞게 구체화할 추상 메서드
    public abstract void displayInfo();
}

// 3. 구체 클래스 1 (전자제품 상품)
class Electronics extends Product {
    private int warrantyMonths;

    public Electronics(String name, double price, int warrantyMonths) {
        super(name, price);
        this.warrantyMonths = warrantyMonths;
    }

    @Override
    public void displayInfo() {
        System.out.printf("[전자제품] %s - 가격: $%.2f (보증기간: %d개월)%n", 
                getName(), getPrice(), warrantyMonths);
    }
}

// 4. 구체 클래스 2 (의류 상품)
class Clothing extends Product {
    private String size;

    public Clothing(String name, double price, String size) {
        super(name, price);
        this.size = size;
    }

    @Override
    public void displayInfo() {
        System.out.printf("[의류] %s - 가격: $%.2f (사이즈: %s)%n", 
                getName(), getPrice(), size);
    }
}

// 5. 인터페이스 구현 클래스 1 (신용카드 결제)
class CreditCardPayment implements PaymentMethod {
    private String cardNumber;

    public CreditCardPayment(String cardNumber) {
        this.cardNumber = cardNumber;
    }

    @Override
    public void processPayment(double amount) {
        System.out.printf("신용카드(%s)로 $%.2f 결제가 완료되었습니다.%n", 
                cardNumber.substring(cardNumber.length() - 4), amount);
    }

    @Override
    public String getPaymentType() {
        return "신용카드";
    }
}

// 6. 인터페이스 구현 클래스 2 (카카오페이/포인트 결제)
class SimplePayPayment implements PaymentMethod {
    private String userEmail;

    public SimplePayPayment(String userEmail) {
        this.userEmail = userEmail;
    }

    @Override
    public void processPayment(double amount) {
        System.out.printf("간편결제 계정(%s)에서 $%.2f 결제가 완료되었습니다.%n", userEmail, amount);
    }

    @Override
    public String getPaymentType() {
        return "간편결제";
    }
}

// 7. 주문 관리 클래스 (List와 Interface를 관리 및 활용)
class ShoppingCart {
    // Product 객체들을 담는 List
    private List<Product> items = new ArrayList<>();

    public void addProduct(Product product) {
        items.add(product);
        System.out.println("장바구니 추가 완료: " + product.getName());
    }

    public void showCartItems() {
        System.out.println("\n--- 장바구니 목록 ---");
        for (Product item : items) {
            item.displayInfo();
        }
    }

    public double calculateTotal() {
        double total = 0;
        for (Product item : items) {
            total += item.getPrice();
        }
        return total;
    }

    // PaymentMethod 인터페이스를 파라미터로 받아 다형성 활용
    public void checkout(PaymentMethod paymentMethod) {
        double totalAmount = calculateTotal();
        System.out.println("\n--- 결제 진행 ---");
        System.out.printf("결제 수단: %s | 총 금액: $%.2f%n", paymentMethod.getPaymentType(), totalAmount);
        paymentMethod.processPayment(totalAmount);
        items.clear(); // 결제 완료 후 장바구니 비우기
    }
}

// 메인 실행 클래스
// 장바구니 생성
ShoppingCart cart = new ShoppingCart();

// 1. 다양한 Product 객체를 생성하여 List에 추가 (다형성)
cart.addProduct(new Electronics("맥북 프로", 2400.0, 12));
cart.addProduct(new Clothing("오버핏 후드티", 59.9, "L"));
cart.addProduct(new Electronics("무선 마우스", 29.5, 6));

// 2. 장바구니 내역 출력
cart.showCartItems();

// 3. 인터페이스(PaymentMethod)를 구현한 결제 수단으로 결제 실행
PaymentMethod creditCard = new CreditCardPayment("1234-5678-9012-3456");
cart.checkout(creditCard);
```

```java
// 2. 상속과 생성자 호출 순서, 올바른 활용 (주제 4, 8)

class Parent {
    Parent() {
        System.out.println("1. Parent 생성자 호출");
    }
    void display() { System.out.println("Parent Display"); }
}

class Child extends Parent {
    Child() {
        super(); // 생략해도 컴파일러가 자동 삽입하여 부모 생성자가 먼저 호출됨
        System.out.println("2. Child 생성자 호출");
    }
    @Override
    void display() { System.out.println("Child Display (상속의 다형성)"); }
}

// 생성자 호출 순서 확인
Parent obj = new Child(); 
obj.display(); // 오버라이딩된 메서드 호출 확인
```

```java
// 3. 레코드와 클래스의 비교, 생성자, 캡슐화 (주제 5, 7, 10)

// 일반 클래스 (기본적으로 equals를 오버라이딩 하지 않으면 == 와 동일하게 참조 비교)
class StudentClass {
    String name;
    int id;
    StudentClass(String name, int id) { this.name = name; this.id = id; }
}

// 레코드 (불변 데이터 객체. equals, hashCode, toString 자동 생성 및 캡슐화 지원)
record StudentRecord(String name, int id) {
    // 레코드의 컴팩트 생성자 (주제 7 - 속성 검증 등에 활용)
    public StudentRecord {
        if (id < 0) throw new IllegalArgumentException("학번은 양수여야 합니다.");
    }
}

StudentClass c1 = new StudentClass("김도연", 20252183);
StudentClass c2 = new StudentClass("김도연", 20252183);

StudentRecord r1 = new StudentRecord("김도연", 20252183);
StudentRecord r2 = new StudentRecord("김도연", 20252183);

// == (참조 비교)와 equals (값 비교)의 차이 확인
System.out.println("Class == : " + (c1 == c2));               // false
System.out.println("Class equals : " + (c1.equals(c2)));      // false
System.out.println("Record == : " + (r1 == r2));              // false
System.out.println("Record equals : " + (r1.equals(r2)));     // true
```

```java
// 4. 불변/가변, 수정가능/수정불가 리스트 (주제 6)

import java.util.ArrayList;
import java.util.Collections;
import java.util.List;

// 1. 가변(Mutable) 리스트
List<String> mutableList = new ArrayList<>();
mutableList.add("A");
mutableList.add("B");

// 2. 수정불가(Unmodifiable) 뷰
// 원본 mutableList가 변경되면 unmodifiableView의 내용도 변경되지만, 
// unmodifiableView 자체를 통한 add/remove는 불가능
List<String> unmodifiableView = Collections.unmodifiableList(mutableList);

// 3. 불변(Immutable) 리스트 (Java 9+)
// 완전히 복사되어 원본과 독립적이며 수정 절대 불가
List<String> immutableList = List.of("X", "Y");

try {
    unmodifiableView.add("C"); // UnsupportedOperationException 발생
} catch (Exception e) {
    System.out.println("Unmodifiable 에러: " + e.getClass().getSimpleName());
}

try {
    immutableList.add("Z"); // UnsupportedOperationException 발생
} catch (Exception e) {
    System.out.println("Immutable 에러: " + e.getClass().getSimpleName());
}
```

```java
// 5. 함수 파라미터 재지정과 수정 (주제 9)

class ParameterTest {
    // primitive 타입과 reference 타입의 매개변수 동작 차이
    static void testModification(int primitiveVal, int[] arrayRef) {
        // 재지정(Reassignment): 로컬 변수가 다른 곳을 가리키게 됨 (원본 영향 없음)
        primitiveVal = 100; 
        
        // 수정(Modification): 참조된 객체의 내부 상태를 변경 (원본 영향 있음)
        arrayRef[0] = 999;  
        
        // 참조 자체를 재지정 (원본 영향 없음)
        arrayRef = new int[]{1, 2, 3}; 
    }
}

int a = 10;
int[] arr = {10, 20, 30};

ParameterTest.testModification(a, arr);

System.out.println("Primitive 원본: " + a);       // 10 (변경 안 됨)
System.out.println("Array 원본[0]: " + arr[0]);   // 999 (수정 됨)
```

```java
public interface Animal {
    // 상수 (public static final 생략 가능)
    String PLANET = "Earth";

    // 추상 메서드 (public abstract 생략 가능)
    void sound();

    // 디폴트 메서드 (Java 8 이상, 필요 시 오버라이드 가능)
    default void sleep() {
        System.out.println("Zzz...");
    }
}

// implements 키워드를 사용하여 인터페이스를 구현합니다.
public class Dog implements Animal {
    @Override
    public void sound() {
        System.out.println("멍멍!");
    }
}

public class Cat implements Animal {
    @Override
    public void sound() {
        System.out.println("야옹~");
    }
}

Animal myDog = new Dog();
Animal myCat = new Cat();
myDog.sound(); // 출력: 멍멍!
myCat.sound(); // 출력: 야옹~
myDog.sleep(); // 출력: Zzz... (디폴트 메서드 호출)
```
